```python
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path
from data_loader import load_datasets
```

```python
base_path = Path(r"C:.\1.NoAuction_Zscore")

x_train, y_train, x_val, y_val, x_test, y_test = load_datasets(base_path=base_path, timestamp_per_sample=100)

# x: (n_samples, temporal lookback, LOB features, channel for CNN)
# y: (n_samples, classes)
print(x_train.shape, y_train.shape) 
print(x_val.shape, y_val.shape)
print(x_test.shape, y_test.shape)
#80_000 samples of shape (10x40x1); 80_000 x3 labels
```

```python
classes = np.argmax(y_train, axis=1)

plt.hist(classes, bins=np.arange(y_train.shape[1] + 1) - 0.5, edgecolor="black")
plt.xticks(range(y_train.shape[1]))
plt.title("classes distribution")
plt.show()
```

```python
i = 0 

sample = x_train[i, :, :, 0]  # shape: (100, 40)

# prices
plt.figure(figsize=(12, 4))
plt.plot(sample[:, 0], label="ask price L1")
plt.plot(sample[:, 2], label="bid price L1")
plt.plot((sample[:, 0]+sample[:, 2])/2, label="MIDPRICE")
plt.title(f"Prices - sample {i}")
plt.grid(True)
plt.legend()
plt.show()

# volumes
plt.figure(figsize=(12, 4))
plt.plot(sample[:, 1], label="ask volume L1")
plt.plot(sample[:, 3], label="bid volume L1")
plt.title(f"Volumes - sample {i}")
plt.grid(True)
plt.legend()
plt.show()
```
